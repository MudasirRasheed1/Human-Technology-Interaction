```python
import os

def save_tree(root_dir, output_file):
    with open(output_file, "w", encoding="utf-8") as f:
        for root, dirs, files in os.walk(root_dir):
            # Calculate indentation based on folder depth
            level = root.replace(root_dir, "").count(os.sep)
            indent = "    " * level

            # Folder name
            f.write(f"{indent}📁 {os.path.basename(root)}/\n")

            # Files
            sub_indent = "    " * (level + 1)
            for file in files:
                f.write(f"{sub_indent}📄 {file}\n")


# Folder whose structure you want to save
root_directory = r"C:\Users\User\Desktop\VS_CODE\Sahi\AMIGOS PreProcessed"

# Output text file
output_file = r"C:\Users\User\Desktop\VS_CODE\Sahi\AMIGOS_PreProcessed.txt"

save_tree(root_directory, output_file)

print(f"Tree structure saved to: {output_file}")
```
